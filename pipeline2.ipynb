```python
def download_and_transcribe_videos(video_id):
    output_folder = "D:\Karya files\Audio_outputs"
    csv_file = "video_metadata.csv"
    # csv_file2 = "video_metadata_2.csv"
    language='Bhojpuri'


    url = "https://www.youtube.com/watch?v=" + video_id
    print("Downloading from URL:", url)

    try:
        get_audio(video_id, output_folder, csv_file, language)
    except Exception as e:
        print("Error downloading audio:", e)
        return
    
    try:
        diarization= diarize_audio(video_id,output_folder)
    except Exception as e:
        print("Error during diarization:", e)
        return
    
    try:
        transcript = generate_transcript(video_id,output_folder)
    except Exception as e:
        print("Error generating transcript:", e)
        transcript = None

    try:
        add_transcript_to_metadata(video_id, transcript, diarization,csv_file)
    except Exception as e:
        print("Error updating metadata:", e)
```

```python
# download_and_transcribe_videos(video_id)
for video in video_ids:
    download_and_transcribe_videos(video)
```
